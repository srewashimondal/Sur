```python
genre_summary = (
    df.groupby("label")
      .agg(
          song_count=("file_name", "count"),
          average_tempo=("tempo", "mean"),
          median_tempo=("tempo", "median"),
          average_rmse=("rmse", "mean"),
          average_spectral_centroid=("spectral_centroid", "mean")
      )
      .round(2)
      .sort_values("song_count", ascending=False)
)

genre_summary
```

```python
df["label"].value_counts().plot(
    kind="bar",
    title="Number of Songs by Genre",
    xlabel="Genre",
    ylabel="Song Count"
)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
tempo_by_genre = (
    df.groupby("label")["tempo"]
      .mean()
      .sort_values(ascending=False)
)

tempo_by_genre.plot(
    kind="bar",
    title="Average Tempo by Genre",
    xlabel="Genre",
    ylabel="Average BPM"
)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
rmse_by_genre = (
    df.groupby("label")["rmse"]
      .mean()
      .sort_values(ascending=False)
)

rmse_by_genre.plot(
    kind="bar",
    title="Average RMSE by Genre",
    xlabel="Genre",
    ylabel="Average RMSE"
)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
def tempo_category(tempo):
    if tempo < 90:
        return "Slow"
    elif tempo < 120:
        return "Moderate"
    elif tempo < 140:
        return "Fast"
    else:
        return "Very Fast"

df["tempo_category"] = df["tempo"].apply(tempo_category)
```

```python
pd.crosstab(df["label"], df["tempo_category"])
```

Band and Rabindra contain many fast songs.
Hip-hop is mostly moderate tempo in this dataset.
Nazrul has very few slow songs.
Rabindra has the most very-fast songs.

```python
# Convert the counts to percentages
tempo_percent = pd.crosstab(
    df["label"],
    df["tempo_category"],
    normalize="index"
).mul(100).round(1)

tempo_percent

# Find 10 fastest songs
df[["file_name", "label", "tempo"]].sort_values(
    "tempo",
    ascending=False
).head(10)
```
